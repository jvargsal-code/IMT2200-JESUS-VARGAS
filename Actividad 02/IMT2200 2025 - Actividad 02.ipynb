## 1. Extraer datos

Utilizando el URL indicado en el enunciado, obtenga la información de los primeros 151 pokémones y guárdelos en una lista llamada `pokemones`. 

Para ello debe usar un `for` que itere sobre los diferentes IDs de pokemon y rescate la información de cada uno de ellos, concatenándola en una lista. Antes de iterar sobre los 151 pokemones, pruebe con 1 o 2 y una vez que verifique que su código está funcionando, aplique a los 151 primeros pokemones.

````python
import requests as req
````

````python
pokemones = [] 

for i in range(1, 152):  # Pokemones del 1 al 151
    url = f'https://pokeapi.co/api/v2/pokemon/{i}'
    response = req.get(url)
    data = response.json()

    name = data['name']

    # Tipos
    types = data['types']
    type1 = types[0]['type']['name']
    type2 = types[1]['type']['name'] if len(types) > 1 else None

    # Stats
    stats = {stat['stat']['name']: stat['base_stat'] for stat in data['stats']}

    pokemones.append({
        'name': name,
        'type1': type1,
        'type2': type2,
        'hp': stats['hp'],
        'attack': stats['attack'],
        'defense': stats['defense'],
        'special-attack': stats['special-attack'],
        'special-defense': stats['special-defense'],
        'speed': stats['speed'],
        "total_stats": 
    })
````

#### 4.1 ¿Cuál son los 3 tipos (*water, fire, ground,* etc.) más comunes en los pokémones de la primera generación?

````python
#chatgpt

# Juntar los tipos de las dos columnas en una sola serie 
all_types = pd.concat([df['type1'], df['type2']])

# Eliminar valores vacíos (None)
all_types = all_types.dropna()

# Contar frecuencia de cada tipo
type_counts = all_types.value_counts()

# Mostrar los 3 tipos más comunes
print("Los 3 tipos de Pokémon más comunes son:")
print(type_counts.head(3))
````

El tipo más común es: Veneno, Agua y Normal.

#### 4.2 ¿Cuál es el pokémon más rápido? ¿Y el más lento?

````python
# Pokémon más rápido

fastest = df.loc[df['speed'].idxmax()]
print("Pokémon más rápido:")
print(f"{fastest["name"]} Velocidad: {fastest["speed"]}")


# Pokémon más lento
slowest = df.loc[df['speed'].idxmin()]
print("Pokémon más lento:")
print(f"{slowest["name"]} Velocidad: {slowest["speed"]}")
````
